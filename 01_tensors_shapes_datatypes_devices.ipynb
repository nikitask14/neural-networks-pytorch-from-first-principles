For one neuron in a layer of a NN, we wrote weighted sum like:

$$ \mathbf z=W\mathbf x+\mathbf b $$

That equation requires us to store and manipulate a lot of numbers:

$$ \mathbf x= \begin{bmatrix} x_1\\x_2\\x_3 \end{bmatrix} $$

and perhaps

$$ W= \begin{bmatrix} w_{11}&w_{12}&w_{13}\\ w_{21}&w_{22}&w_{23} \end{bmatrix} $$

A neural network is therefore doing huge amounts of numerical computation with vectors, matrices, and eventually higher-dimensional collections of numbers.

Ordinary Python can store numbers

```python
[1, 2, 3]
```

But a normal Python list was not designed specifically for fast linear algebra, GPU computation, or neural-network gradient calculations.

That is where PyTorch enters. PyTorch provides the machinery needed for Neural Network computations.

***dtype*** matters because it controls things like:

1. how much memory each value uses
2. what range/precision the values can represent
3. which mathematical operations make sense

For neural networks, the most important point is this:

Weights and gradients usually need floating-point numbers, because learning depends on small continuous changes like \(0.003\), \(1.27\), \(-0.0004\), not just whole numbers.

```python
x = torch.tensor([1 ,2 ,3])
y = torch.tensor([1.0, 2.0, 3.0])
print(x.dtype)
print(y.dtype)
```

***torch.int64*** means integer values represented using 64 bits.

***torch.float32*** means floating-point values represented using 32 bits.

shape = how the values are arranged
	​

dtype = what kind of numerical values are stored
	​


```python
x = torch.tensor([1, 2, 3], dtype = torch.float32)
```

Even though the values look like integers, PyTorch stores them as floating-point numbers.

Why do we need this?

For example, imagine these are feature values:

[18, 25, 40]

They may represent ages, so whole numbers make sense as raw data.

But once those values enter a neural network, they may be multiplied by weights like:

$$ 0.37,\quad -0.12,\quad 1.08 $$

So we may explicitly tell PyTorch:

“Store these input values as floating-point numbers because they are going into floating-point computation.”

That is what dtype = torch.float32 gives us control over how the tensor is represented, independent of how the original numbers were written.

```python
ages = torch.tensor([18, 25, 40], dtype = torch.float32)
print(ages)
print(ages.dtype)
```

```python
ages.device
```

A tensor’s device tells PyTorch where that tensor is physically stored and where calculations involving it should be carried out.

So if a tensor lives on the CPU:

$$ \text{tensor} \rightarrow \text{CPU computation} $$

If it lives on a GPU:
$$ \text{tensor} \rightarrow \text{GPU computation} $$

Why is GPU useful for neural networks? Because NN training involves enormous numbers of very similar numerical operations — especially matrix multiplications. GPUs are designed to perform many such operations in parallel, so they can be much faster.

There is one very important practical rule:

Tensors involved in the same operation generally need to be on the same device.

```python
torch.cuda.is_available()
```

So read that line as:

“PyTorch, is a CUDA-capable GPU available to me?”

The next device idea is simply moving a tensor.

```python
ages.to("cpu")
ages.to("cuda")
```
