```python
import torch
import torch.nn as nn
```

```python
class MyModel(nn.Module):
  def __init__(self):
    super().__init__()

    self.layer1 = nn.Linear(4,6)
    self.layer2 = nn.Linear(6,3)
    self.layer3 = nn.Linear(3,1)

# self gives forward() access to it's model's layers.

  def forward(self, x):
    x = self.layer1(x)
    print("After layer1:", x.shape)

    x = self.layer2(x)
    print("After layer 2:", x.shape)

    x = self.layer3(x)
    print("After layer 3:", x.shape)
    return x

model = MyModel()
print(model)

print()
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
output = model(x)

print("Input:", x)
print("Input shape:", x.shape)
print()

print("Output:", output)
print("Output Shape:", output.shape)
print()

y = torch.tensor([2.0])

print("Prediction:", output)
print("Target:", y)

print()
print("Prediction Shape",output.shape)
print("Target Shape:", y.shape)

loss_fn = nn.MSELoss()

loss = loss_fn(output, y)

print()
print("Prediciton:", output)
print("Target:", y)
print("Loss:", loss)

#Use SGD to update all the model's weight and baises with learning rate
0.01
optimizer = torch.optim.SGD(model.parameters(), lr = 0.01)
print(model.layer1.weight[0][0])

# for all the trainable weights and biases that contributed to the loss,
# and stores those gradients in their .grad.
loss.backward()

# Use the gradients we just computed and update the weights and biase
optimizer.step()
print(model.layer1.weight[0][0])
print(model.layer1.weight.grad[0][0])

# clear the old gradients before computing new ones
optimizer.zero_grad()

print(model.layer1.weight.grad)
```
