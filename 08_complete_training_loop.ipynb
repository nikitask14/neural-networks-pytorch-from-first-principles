```python
import torch
import torch.nn as nn
```

```python
# Define the Neural Netwwork
class MyModel(nn.Module):
  def __init__(self):
    super().__init__()


    # Three linear layers:
    # 4 input features → 6 neurons → 3 neurons → 1 output
    self.layer1 = nn.Linear(4,6)
    self.layer2 = nn.Linear(6,3)
    self.layer3 = nn.Linear(3,1)

  #Pass data x through the model and pyTorch then invokes model's
  # fwd computation.
  def forward(self, x):
    x = self.layer1(x)
    x = self.layer2(x)
    x = self.layer3(x)
    return x

#Create an instance of the MyModel class called model.
model = MyModel()
print(model)

#x contains 4 training examples, each with 4 input features.
x = torch.tensor([
    [1., 2., 3., 4.],
    [2., 1., 0., 3.],
    [4., 2., 1., 5.],
    [3., 3., 2., 1.]
], dtype=torch.float32)

print(x.shape)

# y contains the correct target value for each training example.
# There are 4 examples and 1 target value per example.
y = torch.tensor([
    [10.],
    [6.],
    [12.],
    [9.]
], dtype=torch.float32)

print(y.shape)

print("Input tensor:", x)
# print("Predicted value:", output)
print("Target value:", y)

# nn.MSELoss() itself does not change from epoch to epoch
# so we don't need to recreate it 100 times.
# we define it once.
# and returns their mean as one scalar loss.
loss_fn = nn.MSELoss()


#Next we tell pyTorch what the model parameters that need to be updated
# and the learning rate.
# So that it is ready when the data arrives.
# lr specifies the learning rate used during parameter updates.
optimiser = torch.optim.SGD(model.parameters(), lr = 0.01)

# Repeat the training process for 100 epochs.
# In this example, the whole dataset is used in every epoch.
for epoch in range(100):

  # clear the old gradients from previous epoch before computing new ones
  # PyTorch accumulates gradients unless we explicitly clear them.
  optimiser.zero_grad()

  # Forward pass:
  # pass all 4 training examples through the model
  # using the model's current weights and biases.
  output = model(x)

  # Compare the 4 predictions with the 4 target values
  # and compute the mean squared error.
  loss = loss_fn(output, y)


  #compute the gradient of loss wrt to all learning parameters.
  loss.backward()


  #update the weights and baisies using those gradients
  # and the learning rate.
  optimiser.step()

  if epoch % 10 == 0:
    print(f"Epoch:{epoch},Loss: {loss.item()} ")





```
