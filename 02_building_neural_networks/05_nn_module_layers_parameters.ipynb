```python
import torch
import torch.nn as nn
```

#### **A small neural network**

We will represent the following architecture:

3 → 5 → 2 → 1

This means:
- 3 input features
- first layer produces 5 outputs
- second layer produces 2 outputs
- final layer produces 1 output

Each transition will be represented using 'nn.Linear'

---



### **Creating the model container**

We want the three layers to belong to one neural-network model.

'nn.Module' provides the basic PyTorch machinery for neural-network models, so we create our own model class by inheriting from it.

```python
class MyModel(nn.Module):
  def __init__(self):
    super().__init__()

    self.layer1 = nn.Linear(3,5)
    self.layer2 = nn.Linear(5,2)
    self.layer3 = nn.Linear(2,1)
```

Read it as:

1. class MyModel(nn.Module): → create our own model type using nn.Module
2. __init__ → defines how a MyModel object will be set up
3. super().__init__() → initialize the inherited nn.Module machinery first
4. We use self.layer1, self.layer2, etc. because we want these layers to belong to each MyModel object.

#### **Creating the model object**

'MyModel' is the blueprint.

Now we create one actual model object from that blueprint.
When 'MyModel()' is called, '__init__()' runs and the three 'nn.Linear' layers are created.

```python
model = MyModel()
print(model)
```

```python
model.layer1.weight
```

```python
model.layer1.bias
```

```python
print(model.layer1.weight.shape)
print(model.layer2.weight.shape)
print(model.layer3.weight.shape)
```

For:

nn.Linear(3, 5)

We might have expect the weight shape to be:

$$ [3,5] $$

But PyTorch stores it as:

$$ [5,3] $$

because it organizes the weights as:

**one row per output neuron**

So for layer1:

5 neurons


each neuron has 3 weights

therefore:

$$ \boxed{\text{weight shape}=[5,3]} $$

```python
print(model.layer1.bias.shape)
print(model.layer2.bias.shape)
print(model.layer3.bias.shape)
```

A useful rule to keep:

$$ \boxed{\texttt{nn.Linear(in, out).weight.shape}=[out,in]} $$ $$ \boxed{\texttt{bias.shape}=[out]} $$

```python
for name, parameter in model.named_parameters():
    print(name, parameter.shape)
```

```python
for parameter in model.parameters():
  print(parameter.shape)
```

```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
  def __init__(self):
    super().__init__()

    self.layer1 = nn.Linear(4,6)
    self.layer2 = nn.Linear(6,3)
    self.layer3 = nn.Linear(3,1)

model = MyModel()
print(model)

for name,parameter in model.named_parameters():
  print(name, parameter.shape)


```

**model.named_parameters()** returns each trainable parameter with its name

**while model.parameters()** returns only the parameter tensors themselves.
