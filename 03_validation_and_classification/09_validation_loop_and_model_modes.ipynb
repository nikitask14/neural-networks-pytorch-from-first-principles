```python
import torch
import torch.nn as nn
```

```python

class MyModel(nn.Module):
  def __init__(self):
    super().__init__()

    self.layer1 = nn.Linear(4,6)
    self.layer2 = nn.Linear(6,3)
    self.layer3 = nn.Linear(3,1)

  def forward(self, x):
    x = self.layer1(x)
    x = self.layer2(x)
    x = self.layer3(x)

    return x

model = MyModel()
loss_fn = nn.MSELoss()
optimiser = torch.optim.SGD(model.parameters(), lr = 0.01)

x_train = torch.tensor([
    [1, 2, 3, 4],
    [2, 1, 0, 3],
    [4, 2, 1, 5],
    [3, 3, 2, 1]
], dtype = torch.float32)

y_train = torch.tensor([
    [10],
    [6],
    [12],
    [9]
], dtype = torch.float32)

x_val = torch.tensor([
    [1, 1, 2, 2],
    [5, 2, 1, 3]
], dtype = torch.float32)

y_val = torch.tensor([
     [6],
      [11]

], dtype = torch.float32)

print(x_train.shape)
print(y_train.shape)
print(x_val.shape)
print(y_val.shape)

for epoch in range(100):
  optimiser.zero_grad()
  model.train()
  train_output = model(x_train)
  loss = loss_fn(train_output,y_train )
  loss.backward()
  optimiser.step()

  model.eval()
  with torch.no_grad():
    val_output = model(x_val)
    val_loss = loss_fn(val_output, y_val)

  if epoch % 10 == 0:
    print(f"Epoch:{epoch}, Training loss:{loss.item()}")
    print(f"Epoch:{epoch}, Validation Loss:{val_loss.item()}")
    print()

```
